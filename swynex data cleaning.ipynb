```python
# Convert order_date in datetime
df["order_date"] = pd.to_datetime(
    df["order_date"],
    format="mixed",
    errors="coerce"
)
```

```python
# check datatype
print(df["order_date"])
print(df["order_date"].dtype)
```

```python
# Display order_date
df["order_date_display"] = df["order_date"].dt.strftime("%d/%m/%y")

print(df["order_date_display"].head(10))
```

```python
#Check columns and  their data types
print(df.columns)
print(df.dtypes)
```

```python
# Invalid values
#Check quantity
print(df[df["quantity"] <= 0])
```

```python
#Check unit_price
print(df[df["unit_price"] <= 0])
```

```python
#Check discount
print(df[(df["discount"] < 0) | (df["discount"] > 1)])
```

```python
# Check revenue
print(df[df["revenue"] < 0])
```

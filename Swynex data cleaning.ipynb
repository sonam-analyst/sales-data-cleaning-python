```python
# Convert order_date in datetime
df["order_date"] = pd.to_datetime(
    df["order_date"].str.replace("-", "/"),
    format="%m/%d/%Y",
    errors="coerce"
)
```

```python
#Check spaces in text
text_columns = [
    "product_category",
    "region",
    "payment_method"
]

for col in text_columns:
    print(col)
    print(df[col].unique())

    for col in text_columns:
        df[col] = df[col].str.strip()

```

```python
#Validate revenue
df["calculated_revenue"] = (
    df["quantity"] *
    df["unit_price"] *
    (1 - df["discount"])
)
```

```python
df["revenue_difference"] = (
    df["revenue"] -
    df["calculated_revenue"]
)
```

```python
print(
    df[abs(df["revenue_difference"]) > 0.01]
    [["quantity",
      "unit_price",
      "discount",
      "revenue",
      "calculated_revenue",
      "revenue_difference"]]
)
```

```python
df.drop(
    columns=["calculated_revenue", "revenue_difference"],
    inplace=True
)
```

```python
#Final check
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())

```

```python
#export cleane data
df.to_csv("cleaned_data.csv", index=False)

cleaned_df = pd.read_csv("cleaned_data.csv")

print(cleaned_df.head(10))
```
